# 📊 Marketing Funnel & Conversion Analysis

## 🎯 Objective

The goal of this analysis is to evaluate marketing funnel performance and identify where users drop off in the conversion process.

This analysis helps businesses understand how to improve conversion rates and optimize marketing strategies.


## 📥 Data Creation

In this step, we simulate a marketing funnel dataset representing users moving through different stages:

Visitors → Leads → Customers

This allows us to analyze conversion rates and drop-off points.


```python
import pandas as pd

# Create simple funnel dataset
funnel = pd.DataFrame({
    'Stage': ['Visitors', 'Leads', 'Customers'],
    'Count': [10000, 2500, 400]
})

funnel
```

## 📊 Conversion Rate Analysis

We calculate the conversion rates between each stage of the funnel to understand performance.


```python
visitor_to_lead = 2500 / 10000
lead_to_customer = 400 / 2500

print("Visitor → Lead:", visitor_to_lead)
print("Lead → Customer:", lead_to_customer)
```

### 🔍 Insights

- Only 25% of visitors become leads
- Only 16% of leads convert into customers

This indicates a significant drop-off in the funnel.


## 📢 Channel Performance Analysis

We simulate marketing channels to understand which ones generate higher conversion rates.


```python
import numpy as np

np.random.seed(42)

data = pd.DataFrame({
    'Channel': np.random.choice(['Facebook', 'Google Ads', 'Email'], 5000),
    'Stage': np.random.choice(['Visitor', 'Lead', 'Customer'], 5000, p=[0.7, 0.2, 0.1])
})

data.head()
```

## 📈 Conversion by Channel

We analyze how each marketing channel performs across funnel stages.


```python
pd.crosstab(data['Channel'], data['Stage'], normalize='index') * 100
```

### 🔍 Insights

- Some channels generate more leads but fewer customers
- Others may produce fewer leads but higher-quality conversions

This helps identify high-performing marketing channels.


## 📤 Export Data

We export the dataset to Power BI for visualization and dashboard creation.


```python
data.to_csv("../data/funnel_data.csv", index=False)
funnel.to_csv("../data/funnel_summary.csv", index=False)
```
