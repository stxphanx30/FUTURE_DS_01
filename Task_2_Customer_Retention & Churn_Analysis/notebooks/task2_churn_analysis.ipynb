## 📥 Data Loading

In this step, we load the Telco Customer Churn dataset and preview its structure to begin the analysis.


```python
import pandas as pd

# Load dataset
data = pd.read_csv("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")

# Preview
data.head()
```

## 🧐 Data Overview

We examine the structure and data types of the dataset to understand available features.


```python
data.info()
```

## 🎯 Churn Distribution

We analyze the distribution of the churn variable to understand how many customers leave the service.


```python
data['Churn'].value_counts()
```

```python
data['Churn'].value_counts(normalize=True) * 100
```

```python
data['TotalCharges'].head()
```

```python
data.dtypes
```

```python
# Convert TotalCharges properly
data['TotalCharges'] = pd.to_numeric(data['TotalCharges'], errors='coerce')
```

```python
data.isnull().sum()
```

```python
data = data.dropna(subset=['TotalCharges'])
```

```python
data.dtypes
```

## 🎯 Churn Overview

We analyze how many customers have churned versus those who stayed.


```python
data['Churn'].value_counts()
```

```python
data['Churn'].value_counts(normalize=True) * 100
```

## 📄 Churn by Contract Type

We analyze how contract type affects customer churn.


```python
pd.crosstab(data['Contract'], data['Churn'], normalize='index') * 100
```

## 💳 Churn by Payment Method

We analyze whether payment methods influence churn behavior.


```python
pd.crosstab(data['PaymentMethod'], data['Churn'], normalize='index') * 100
```

## 💰 Monthly Charges vs Churn

We analyze whether higher monthly charges are associated with higher churn.


```python
data.groupby('Churn')['MonthlyCharges'].mean()
```

## ⏳ Tenure vs Churn

We analyze how long customers stay before churning.


```python
data.groupby('Churn')['tenure'].mean()
```

## ⏳ Customer Retention Segments

We group customers based on their tenure to understand retention patterns over time.


```python
# Create tenure groups
bins = [0, 12, 24, 48, 72]
labels = ['0-1 Year', '1-2 Years', '2-4 Years', '4-6 Years']

data['TenureGroup'] = pd.cut(data['tenure'], bins=bins, labels=labels)
```

## 📉 Churn by Tenure Group

We analyze churn rates across different customer lifetime stages.


```python
pd.crosstab(data['TenureGroup'], data['Churn'], normalize='index') * 100
```

## 💰 Customer Lifetime Value (Basic Estimate)

We estimate customer value based on tenure and monthly charges.


```python
data['CLV'] = data['tenure'] * data['MonthlyCharges']

data.groupby('Churn')['CLV'].mean()
```

```python
data.groupby('TenureGroup')['Churn'].value_counts(normalize=True) * 100
```
